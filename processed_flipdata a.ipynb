```python

import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


data = df.copy()

price_col = next(
    (c for c in data.columns if c.strip().lower() == "price"),
    None
)

if price_col is None:
    raise ValueError("Price column not found in df.")

data[price_col] = pd.to_numeric(
    data[price_col].astype(str)
    .str.replace(r"[₹,\s]", "", regex=True),
    errors="coerce"
)

data = data.dropna(subset=[price_col]).drop_duplicates()

X = data.drop(columns=[price_col])
y = data[price_col]

# Remove columns that contain no useful values
X = X.dropna(axis=1, how="all")

numeric_features = X.select_dtypes(
    include=["number", "bool"]
).columns.tolist()

categorical_features = [
    c for c in X.columns if c not in numeric_features
]

# 2. Preprocess all available features
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 4. Train the improved model
full_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2,
        n_jobs=-1
    ))
])

full_model.fit(X_train, y_train)

y_pred_full = full_model.predict(X_test)

mae_full = mean_absolute_error(y_test, y_pred_full)
mse_full = mean_squared_error(y_test, y_pred_full)
rmse_full = np.sqrt(mse_full)
r2_full = r2_score(y_test, y_pred_full)

evaluation_final = pd.DataFrame([{
    "Model": "Random Forest - All Features",
    "MAE": mae_full,
    "MSE": mse_full,
    "RMSE": rmse_full,
    "R2 Score": r2_full,
    "Training Rows": len(X_train),
    "Testing Rows": len(X_test),
    "Features Used": X.shape[1]
}])

print("FINAL MODEL EVALUATION")
display(evaluation_final)


feature_names = (
    full_model.named_steps["preprocessor"]
    .get_feature_names_out()
)

importances = (
    full_model.named_steps["regressor"].feature_importances_
)

feature_importance_final = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values("Importance", ascending=False)

print("TOP 15 IMPORTANT FEATURES")
display(feature_importance_final.head(15))


joblib.dump(full_model, "mobile_price_model_final.joblib")

evaluation_final.to_csv(
    "final_model_evaluation.csv", index=False
)

feature_importance_final.to_csv(
    "final_feature_importance.csv", index=False
)

predictions_final = pd.DataFrame({
    "Actual Price": y_test.to_numpy(),
    "Predicted Price": y_pred_full
})

predictions_final.to_csv(
    "final_price_predictions.csv", index=False
)

print("\nPROJECT OUTPUT FILES SAVED:")
print("1. mobile_price_model_final.joblib")
print("2. final_model_evaluation.csv")
print("3. final_feature_importance.csv")
print("4. final_price_predictions.csv")
print("\nFinal model training and evaluation completed.")
```
